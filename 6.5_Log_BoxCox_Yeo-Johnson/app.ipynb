# Log Transformation

```python
from sklearn.preprocessing import PowerTransformer
import numpy as np
from scipy.stats import skew
import matplotlib.pyplot as plt

X_train = [[10], [20], [30], [40], [60], [100], [200], [500], [1000], [5000]]
X_test  = [[50], [80], [150], [300], [800]]

X_train_log = np.log1p(X_train)
X_test_log = np.log1p(X_test)
print("Before Log Transformation :")
print(X_train)

print("After Log Transformation :")
print(X_train_log)

print("Before Log Transformation :")
print(X_test)

print("After Log Transformation :")
print(X_test_log)

```

# Box-Cox Transformation

```python
from sklearn.preprocessing import PowerTransformer
import numpy as np
import matplotlib.pyplot as plt

X_train = [[10], [20], [30], [40], [60], [100], [200], [500], [1000], [5000]]
X_test  = [[50], [80], [150], [300], [800]]

boxcox = PowerTransformer(method="box-cox")

X_train_boxcox = boxcox.fit_transform(X_train)
X_test_boxcox = boxcox.transform(X_test)

print("Before Log Transformation :")
print(X_train)

print("After Log Transformation :")
print(X_train_boxcox)

print("Before Log Transformation :")
print(X_test)

print("After Log Transformation :")
print(X_test_boxcox)

```

```python
from scipy.stats import skew

print("Original:", skew(X_train))
print("Log:", skew(X_train_log.flatten()))
print("Box-Cox:", skew(X_train_boxcox.flatten()))
```
